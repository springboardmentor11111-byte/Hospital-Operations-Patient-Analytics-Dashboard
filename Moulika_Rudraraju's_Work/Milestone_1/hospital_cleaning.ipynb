# Hospital Data Cleaning
Professional Pandas data cleaning notebook.

```python
import pandas as pd
import numpy as np

df=pd.read_csv("/home/rguktongole/Downloads/infosysspringboard/hospital_raw_data.csv")
df.head()
```

```python
# Basic inspection
print(df.shape)
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
# Remove duplicate rows
df=df.drop_duplicates()

# Trim whitespace from object columns
for c in df.select_dtypes(include='object').columns:
    df[c]=df[c].astype(str).str.strip()

# Replace blank strings with NaN
df.replace(r'^\s*$', np.nan, regex=True, inplace=True)

# Fill missing values
for c in df.columns:
    if df[c].dtype=='object':
        mode=df[c].mode(dropna=True)
        if not mode.empty:
            df[c]=df[c].fillna(mode[0])
    else:
        df[c]=df[c].fillna(df[c].median())

# Standardize text
for c in df.select_dtypes(include='object').columns:
    df[c]=df[c].str.title()
```

```python
# Attempt date conversion
for c in df.columns:
    if 'date' in c.lower():
        df[c]=pd.to_datetime(df[c], errors='coerce')
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
df.to_csv('/home/rguktongole/Downloads/infosysspringboard/hospital_cleaned.csv',index=False)
print('Saved hospital_cleaned.csv')
```
